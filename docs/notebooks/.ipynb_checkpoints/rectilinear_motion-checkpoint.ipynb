```python
import numpy as np

from matplotlib import pyplot as plt
```

```python
def get_pos(t, x0, v):
    
    c = v * t
    
    x = x0 + c
    
    return x
```

```python
for ii in range(9):
    
    tm = 0.5 * ii

    fig, ax = plt.subplots(figsize=(4,4))#, facecolor='beige')

    xb = np.linspace(-20,50,100)
    
    ax.plot(xb, 0*xb, 'k-')
    
    x1 = get_pos(tm, -20, 5.0)
    x2 = get_pos(tm, 40, -10.0)
    
    ax.set_ylim([-5,8])
    
    ax.set_aspect(1.0)

    ax.plot(x1, 0, 'r.', markersize = 15)
    ax.plot(x2, 0, 'b.', markersize = 15)

    ax.text(0, 3, f't = {tm}')

    h = ax.get_yaxis().set_visible(False)
    
    plt.savefig(f'rect_same_dir/rect_{ii:02d}.png', dpi=450)
    
    plt.close('all')
```
