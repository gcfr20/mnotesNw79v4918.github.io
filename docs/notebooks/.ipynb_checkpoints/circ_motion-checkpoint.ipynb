```python
import numpy as np

from matplotlib import pyplot as plt
```

```python
def get_pos(t, r, th_0, v):
    
    c = v * t
    
    th = th_0 - c/r
    
    return th
```

```python
for ii in range(100):
    
    tm = 0.5 * ii

    fig, ax = plt.subplots(figsize=(4,4), facecolor='beige')

    th = np.linspace(0,2*np.pi,100)

    r = 100.0/np.pi

    ax.plot(r*np.cos(th), r*np.sin(th), 'k-')
    ax.set_aspect('equal')

    th1 = get_pos(tm, r, np.pi/2, -20.0)
    th2 = get_pos(tm, r, 0, 10.0)

    ax.plot(r*np.cos(th1), r*np.sin(th1), 'r.', markersize = 15)
    ax.plot(r*np.cos(th2), r*np.sin(th2), 'b.', markersize = 15)

    ax.text(-6, 0, f't = {tm}')

    h = ax.axis('off')
    
    plt.savefig(f'circle_same_dir/circ_{ii:02d}.png', dpi=450)
    
    plt.close('all')
```
