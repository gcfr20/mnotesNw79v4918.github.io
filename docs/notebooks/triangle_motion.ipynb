```python
import numpy as np

from matplotlib import pyplot as plt
```

```python
vt = (1.0) * np.sin(np.pi/6)
vr = -(1.0) * np.cos(np.pi/6)
```

```python
tm = 0.5 * np.arange(21)

r0 = 15.0 / np.sqrt(3.0)

r1 = r0 + vr * tm

th1 = np.pi/2 + (vt/vr) * np.log(1.0 + vr * tm/r0 + 1.0e-12)

th2 = th1 + np.pi * (2.0/3)

th3 = th2 + np.pi * (2.0/3)

x1 = r1 * np.cos(th1)
y1 = r1 * np.sin(th1)

x2 = r1 * np.cos(th2)
y2 = r1 * np.sin(th2)

x3 = r1 * np.cos(th3)
y3 = r1 * np.sin(th3)
```

```python
vr
```

```python
for ii in range(len(tm)):

    fig, ax = plt.subplots(figsize=(4,4))#, facecolor='beige')
    
    ax.set_xlim([-12,12])
    ax.set_ylim([-12,12])
    
    ax.set_aspect(1.0)
    
    ax.axhline(y=0, color='blue', linewidth=0.5)
    ax.axvline(x=0, color='blue', linewidth=0.5)
    
    ax.plot(x1[:ii+1], y1[:ii+1], 'r.-', markersize = 5)
    ax.plot(x2[:ii+1], y2[:ii+1], 'r.-', markersize = 5)
    ax.plot(x3[:ii+1], y3[:ii+1], 'r.-', markersize = 5)

    ax.plot([x1[ii], x2[ii], x3[ii], x1[ii]],
            [y1[ii], y2[ii], y3[ii], y1[ii]], 'k.-', markersize = 10)
    
  
    ax.text(6, 9, f't = {tm[ii]}')
    
    ax.text(x1[0]+0.3, y1[0]+0.3, 'A', fontdict=dict(color='red'))
    ax.text(x2[0]-1.3, y2[0]-0.6, 'B', fontdict=dict(color='red'))
    ax.text(x3[0]+0.5, y3[0]-0.6, 'C', fontdict=dict(color='red'))

    #h = ax.get_yaxis().set_visible(False)
    
    plt.savefig(f'tri_dir/tri_{ii:02d}.png', dpi=450)
    
    plt.close('all')
```
